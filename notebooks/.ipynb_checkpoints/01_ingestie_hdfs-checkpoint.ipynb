# 01 - Ingestie in HDFS

Scop: incarcam fisierul `flights.csv` din `/data` in HDFS si verificam ca Spark il poate citi distribuit.

Nu am folosit Parquet (probleme laptop). Am pastrat raw CSV in HDFS si s-a salvat ulterior CSV curatat.


```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import IntegerType, DoubleType

spark = (
    SparkSession.builder
    .appName("FlightsProject")
    .master("spark://master:7077")
    .config("spark.executor.memory", "1g")
    .config("spark.executor.cores", "1")
    .config("spark.cores.max", "2")
    .config("spark.sql.shuffle.partitions", "4")
    .getOrCreate()
)
```

```python
import pandas as pd

pd.set_option("display.max_columns", 20)
pd.set_option("display.max_colwidth", 40)
pd.set_option("display.width", 140)

def show_small(sdf, n=20):
    return sdf.limit(n).toPandas()
```

## Configurare cai

Datasetul Kaggle trebuie pus in folderul `/data` din containerul master (legat in Docker Compose de folderul local `./data`)


```python
import os
import subprocess

LOCAL_CSV = "/data/flights.csv"
HDFS_RAW_DIR = "/flights/raw"
HDFS_RAW_CSV = "hdfs://master:9000/flights/raw/flights.csv"

print("Local CSV exists:", os.path.exists(LOCAL_CSV), LOCAL_CSV)
```

## Copiere in HDFS

Comenzile HDFS se ruleaza din notebook.


```python
cmds = [
    ["hdfs", "dfs", "-mkdir", "-p", HDFS_RAW_DIR],
    ["hdfs", "dfs", "-rm", "-f", f"{HDFS_RAW_DIR}/flights.csv"],
    ["hdfs", "dfs", "-put", LOCAL_CSV, f"{HDFS_RAW_DIR}/flights.csv"],
    ["hdfs", "dfs", "-ls", HDFS_RAW_DIR],
]

for cmd in cmds:
    print("$", " ".join(cmd))
    result = subprocess.run(cmd, text=True, capture_output=True)
    print(result.stdout)
    if result.returncode != 0 and "No such file" not in result.stderr:
        print(result.stderr)
```

## Citire cu Spark

Spark citeste direct din HDFS. `repartition(4)` foloseste mai bine cei doi workeri (units of parallel work).


```python
df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(HDFS_RAW_CSV)
    .repartition(4)
)

print("Numar coloane:", len(df.columns))
print("Numar randuri:", df.count())
df.printSchema()

preview_cols = [c for c in [
    "year", "month", "day", "carrier", "flight",
    "origin", "dest", "dep_delay", "arr_delay", "air_time", "distance"
] if c in df.columns]

show_small(df.select(*preview_cols), 5)
```

## Concluzie

Fisierul raw este in HDFS si poate fi citit de Spark. Urmatorul notebook face EDA, cleanup si salveaza varianta curatata ca CSV in HDFS.
